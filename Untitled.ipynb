```python
# =============================================================================
# IMPORTS
# =============================================================================

# --- Manipulação numérica e geração de dados ---
import numpy as np
from sklearn.datasets import make_classification

# --- Divisão de dados em treino e teste ---
from sklearn.model_selection import train_test_split

# --- Modelos do Scikit-learn que serão comparados ---
from sklearn.linear_model import Perceptron       # Perceptron clássico
from sklearn.linear_model import SGDClassifier    # SGD com loss 'perceptron', 'hinge' e 'log_loss'

# --- Métricas de avaliação ---
from sklearn.metrics import (
    accuracy_score,         # Acurácia geral
    classification_report,  # Precisão, recall e F1 por classe
    confusion_matrix,       # Matriz de confusão
    ConfusionMatrixDisplay,  # Exibição visual da matriz de confusão
)

# --- PyTorch: framework de deep learning (implementação manual do Perceptron) ---
import torch                    # Tensores e autograd
import torch.nn as nn           # Módulos de redes neurais (camadas, ativações)
import torch.optim as optim     # Otimizadores (SGD, Adam, etc.)

# --- Visualização ---
import matplotlib.pyplot as plt

```

```python

# =============================================================================
# ETAPA 1 — Geração de Dados Sintéticos
# =============================================================================
# Contexto de negócio:
#   Simulamos uma base de clientes de uma instituição financeira.
#   Cada cliente é descrito por duas características numéricas:
#     - Feature 1: renda mensal normalizada
#     - Feature 2: score de crédito normalizado
#   O objetivo é prever se o crédito deve ser APROVADO (1) ou NEGADO (0).
#
# Por que dados sintéticos?
#   make_classification() gera dados controlados e balanceados, ideais para
#   comparar modelos sem o ruído e as particularidades de dados reais.

RANDOM_STATE = 42  # Semente global — use esta constante em todo o projeto
                   # para garantir reprodutibilidade em splits, modelos e plots

X, y = make_classification(
    n_samples=300,       # 300 clientes no dataset
    n_features=2,        # 2 features (renda e score de crédito)
    n_informative=2,     # ambas as features são preditivas da classe alvo
    n_redundant=0,       # sem features redundantes (cópias lineares de outras)
    n_clusters_per_class=1,  # 1 cluster por classe → separação mais linear,
                             # favorece a comparação justa entre Perceptrons
    random_state=RANDOM_STATE,
)

# Verificação rápida dos dados gerados
print(f"Shape de X : {X.shape}")   # esperado: (300, 2)
print(f"Shape de y : {y.shape}")   # esperado: (300,)
print(f"Classes    : {set(y)}")    # esperado: {0, 1}
print(f"Balanço    : {y.mean():.0%} de aprovações")  # idealmente ~50%

```

```python
# =============================================================================
# ETAPA 2 — Divisão Treino / Teste
# =============================================================================
# Por que dividir os dados?
#   Treinar e avaliar no mesmo conjunto gera uma métrica enganosa (overfitting).
#   Reservamos 30% dos dados para teste — dados que NENHUM modelo verá durante
#   o treinamento — para medir a capacidade de generalização de cada abordagem.

# --- 2.1 Split sklearn (arrays NumPy) ---
# Usado diretamente pelo Perceptron e SGDClassifier do scikit-learn
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,          # 30% para teste  → ~90 clientes
                            # 70% para treino → ~210 clientes
    random_state=RANDOM_STATE,  # mesma semente global definida na Etapa 1
    stratify=y,             # mantém a proporção de aprovados/negados em
                            # ambos os conjuntos (evita viés amostral)
)

print("─── Arrays NumPy (sklearn) ───────────────────────────────")
print(f"Treino : X={X_train.shape} | y={y_train.shape} "
      f"| aprovações: {y_train.mean():.0%}")
print(f"Teste  : X={X_test.shape}  | y={y_test.shape}  "
      f"| aprovações: {y_test.mean():.0%}")

# --- 2.2 Conversão para tensores PyTorch ---
# O PyTorch não aceita arrays NumPy diretamente em sua API de nn.Module.
# torch.float32 é o dtype padrão para dados contínuos em redes neurais.
# .view(-1, 1) transforma y de shape (N,) para (N, 1), necessário porque
# nn.BCELoss (Binary Cross-Entropy) espera que input e target tenham
# o mesmo número de dimensões.

X_train_t = torch.tensor(X_train, dtype=torch.float32)            # (210, 2)
X_test_t  = torch.tensor(X_test,  dtype=torch.float32)            # ( 90, 2)

y_train_t = torch.tensor(y_train, dtype=torch.float32).view(-1, 1)  # (210, 1)
y_test_t  = torch.tensor(y_test,  dtype=torch.float32).view(-1, 1)  # ( 90, 1)

print("─── Tensores PyTorch ─────────────────────────────────────")
print(f"X_train_t : {tuple(X_train_t.shape)} | dtype: {X_train_t.dtype}")
print(f"X_test_t  : {tuple(X_test_t.shape)}  | dtype: {X_test_t.dtype}")
print(f"y_train_t : {tuple(y_train_t.shape)} | dtype: {y_train_t.dtype}")
print(f"y_test_t  : {tuple(y_test_t.shape)}  | dtype: {y_test_t.dtype}")

```
